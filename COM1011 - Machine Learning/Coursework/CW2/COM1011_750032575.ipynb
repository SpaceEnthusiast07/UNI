(7) Load the fashion-MNIST test dataset (`t10k-labels-idx1-ubyte.gz` and `t10k-images-idx3-ubyte.gz`), then obtain the image data denoted as `X_test` and the ground truth labels denoted as `y_test`. After that, pre-process `X_test` by scaling the image pixel values to values between 0 and 1. Source code is provided. [0 mark]

[hint: check https://github.com/zalandoresearch/fashion-mnist?tab=readme-ov-file#get-the-data for detailed information about fashion-MNIST dataset]


```python
# answer to (7). You may need to change path for `t10k-labels-idx1-ubyte.gz` and `t10k-images-idx3-ubyte.gz`
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

def load_mnist(path, kind='train'):
    import os
    import gzip
    import numpy as np

    """Load fashion-MNIST data from `path`"""
    labels_path = os.path.join(path, '%s-labels-idx1-ubyte.gz' % kind)
    images_path = os.path.join(path, '%s-images-idx3-ubyte.gz' % kind)

    with gzip.open(labels_path, 'rb') as lbpath:
        labels = np.frombuffer(lbpath.read(), dtype=np.uint8, offset=8)

    with gzip.open(images_path, 'rb') as imgpath:
        images = np.frombuffer(imgpath.read(), dtype=np.uint8, offset=16).reshape(len(labels), 784)

    return images, labels


X_test, y_test = load_mnist('', kind='t10k')

X_test = X_test/255

print(np.max(X_test))
#print(X_train.shape)
#print(y_train.shape)

print(X_test.shape)
print(y_test.shape)
```

(8) Obtain all the test images and their (ground-truth) labels for '3','6','7', respectively, and then rearrange the samples and labels in a random order. [1 mark]

[hint: check https://numpy.org/doc/2.2/reference/generated/numpy.isin.html and https://numpy.org/doc/2.1/reference/random/generated/numpy.random.permutation.html]



```python
# Answer to (8)
# Extract all images corresponding to labels '3','6', and '7'
mask = np.isin(y_test, [3,6,7])

extractedImages = X_test[mask]
extractedLabels = y_test[mask]

shuffledIndeces = np.random.permutation(len(extractedImages))
shuffledImages = extractedImages[shuffledIndeces]
shuffledLabels = extractedLabels[shuffledIndeces]

print(shuffledLabels)
```

```python
# Answer to (11)
from sklearn.metrics import davies_bouldin_score

# Output general message
print("D.B. score for:")

for k in range(2,5):
    kmeans = KMeans(n_clusters=k, random_state=0, n_init="auto")
    clusterLabels = kmeans.fit_predict(shuffledImages)
    
    # Calculate the db score
    dbScore = davies_bouldin_score(shuffledImages, clusterLabels)

    # Output the db score
    print(f"  k = {k}: {dbScore:.6f}")
```

For a Silhouette score, the range is [-1,1], where a score of 1 is perfect clustering, 0 some data points overlap between clusters, and -1 being points are assigned to the wrong cluster. Therefore, a score closer to 1 is preferred. From the calculations, k = 2 produces the best clustering of points, with a silhouette score of 0.275. Yet again, this is not consistent with the ground truth for the number of clusters in the fashion_mnist dataset. (20). Usually, DBSCAN takes longer than K-means to run, and the time it takes to run is affected by the eps parameter. Explain why that is the case. [4 marks]

Answer to (20)

`eps` (epsilon) is the maximum radius around a given point and is used to determine which points are considered its neighbour. DBSCAN clusters points together that are within this distance threshold (bubble). Since eps is the threshold distance, generally the further you travel from a point, the more points you are going to interact with. This means that as the eps value gets larger, the DBSCAN algorithm will have more points to check against, increasing the time taken for the algorihtm to compute the clusters.

```python
# Answer to (23)
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Mask out each 4 and 6 label
maskFor4And6Train = np.isin(y_train, [4,6])
maskFor4And6Test = np.isin(y_test, [4,6])

# Extract labels 4 and 6 and corresponding images
extractedLabelsTrain = y_train[maskFor4And6Train]
extractedLabelsTest = y_test[maskFor4And6Test]
extractedImagesTrain = X_train[maskFor4And6Train]
extractedImagesTest = X_test[maskFor4And6Test]

# Set up the logistic regression model
lRModelFor4And6 = LogisticRegression(random_state=42, solver='sag', max_iter=200)
# Train the model on the extracted data
lRModelFor4And6.fit(extractedImagesTrain, extractedLabelsTrain)

# Make some predictions on the train data
classificationPredictionTrain = lRModelFor4And6.predict(extractedImagesTrain)
# Make some predictions on the test data
classificationPredictionTest = lRModelFor4And6.predict(extractedImagesTest)

# Calculate that train accuracy
trainAccuracy = accuracy_score(extractedLabelsTrain, classificationPredictionTrain)
# Calculate the test accuracy
testAccuracy = accuracy_score(extractedLabelsTest, classificationPredictionTest)

# List to hold the accuracies for different max_iter values
trainAccuracies = [trainAccuracy]
testAccuracies = [testAccuracy]

# Output the two accuracies
print("Classification Accuracies:")
print(f"  Train: {trainAccuracy}")
print(f"  Test: {testAccuracy}")
```

```python
# Answer to (27)
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score, confusion_matrix, precision_recall_fscore_support
from scipy.ndimage import rotate

# Flip the train and test images of class 4 and 6 horizontally
flippedTrainImages = np.array([np.fliplr(image.reshape(28,28)).reshape(-1) for image in extractedImagesTrain])
flippedTestImages = np.array([np.fliplr(image.reshape(28,28)).reshape(-1) for image in extractedImagesTest])

# Set up the two plot figure
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14,6))

# Choose a random image
randomImage = np.random.randint(0, len(extractedImagesTest))

# Display a regular image
ax1.imshow(extractedImagesTest[randomImage].reshape(28,28), cmap='gray')
ax1.set_title("Original Image")

# Display the flipped image
ax2.imshow(flippedTestImages[randomImage].reshape(28,28), cmap='gray')
ax2.set_title("Flipped Image")

plt.show()


# Combine the regular and flipped images into one table (2d array)
augmentedTrainImages = np.vstack((extractedImagesTrain, flippedTrainImages))
augmentedTestImages = np.vstack((extractedImagesTest, flippedTestImages))
# Duplicate the labels array to it is the same length as the augmented images one
augmentedTrainLabels = np.concatenate((extractedLabelsTrain, extractedLabelsTrain))
augmentedTestLabels = np.concatenate((extractedLabelsTest, extractedLabelsTest))

# Loop through each logistic regression model with different max_iter values
for i in [200,300,500]:
    # Set up the logistic regression model
    # Could use the 'lbfgs' solver, which may result in higher accuracies!
    augmentedLRModel = LogisticRegression(max_iter=i, solver='sag', random_state=42)
    # Train the new model on the regular and flipped images
    augmentedLRModel.fit(augmentedTrainImages, augmentedTrainLabels)

    # Predict labels using training data and calculate accuracy
    predictedTrainLabels = augmentedLRModel.predict(augmentedTrainImages)
    trainAccuracy = accuracy_score(augmentedTrainLabels, predictedTrainLabels)
    # Predict labels using testing data
    predictedTestLabels = augmentedLRModel.predict(augmentedTestImages)
    testAccuracy = accuracy_score(augmentedTestLabels, predictedTestLabels)

    print(f"Classification Accuracies for max_iter = {i}")
    print(f"  Train: {trainAccuracy}")
    print(f"  Test: {testAccuracy}")

```
