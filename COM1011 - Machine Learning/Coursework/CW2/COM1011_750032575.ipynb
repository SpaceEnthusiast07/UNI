(7) Load the fashion-MNIST test dataset (`t10k-labels-idx1-ubyte.gz` and `t10k-images-idx3-ubyte.gz`), then obtain the image data denoted as `X_test` and the ground truth labels denoted as `y_test`. After that, pre-process `X_test` by scaling the image pixel values to values between 0 and 1. Source code is provided. [0 mark]

[hint: check https://github.com/zalandoresearch/fashion-mnist?tab=readme-ov-file#get-the-data for detailed information about fashion-MNIST dataset]


```python
# answer to (7). You may need to change path for `t10k-labels-idx1-ubyte.gz` and `t10k-images-idx3-ubyte.gz`
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score

def load_mnist(path, kind='train'):
    import os
    import gzip
    import numpy as np

    """Load fashion-MNIST data from `path`"""
    labels_path = os.path.join(path, '%s-labels-idx1-ubyte.gz' % kind)
    images_path = os.path.join(path, '%s-images-idx3-ubyte.gz' % kind)

    with gzip.open(labels_path, 'rb') as lbpath:
        labels = np.frombuffer(lbpath.read(), dtype=np.uint8, offset=8)

    with gzip.open(images_path, 'rb') as imgpath:
        images = np.frombuffer(imgpath.read(), dtype=np.uint8, offset=16).reshape(len(labels), 784)

    return images, labels


X_test, y_test = load_mnist('', kind='t10k')

X_test = X_test/255

print(np.max(X_test))
#print(X_train.shape)
#print(y_train.shape)

print(X_test.shape)
print(y_test.shape)
```

(8) Obtain all the test images and their (ground-truth) labels for '3','6','7', respectively, and then rearrange the samples and labels in a random order. [1 mark]

[hint: check https://numpy.org/doc/2.2/reference/generated/numpy.isin.html and https://numpy.org/doc/2.1/reference/random/generated/numpy.random.permutation.html]



```python
# Answer to (8)
# Extract all images corresponding to labels '3','6', and '7'
mask = np.isin(y_test, [3,6,7])

extractedImages = X_test[mask]
extractedLabels = y_test[mask]

shuffledIndeces = np.random.permutation(len(extractedImages))
shuffledImages = extractedImages[shuffledIndeces]
shuffledLabels = extractedLabels[shuffledIndeces]

print(shuffledLabels)
```

```python
# Answer to (11)
from sklearn.metrics import davies_bouldin_score

# Output general message
print("D.B. score for:")

for k in range(2,5):
    kmeans = KMeans(n_clusters=k, random_state=0, n_init="auto")
    clusterLabels = kmeans.fit_predict(shuffledImages)
    
    # Calculate the db score
    dbScore = davies_bouldin_score(shuffledImages, clusterLabels)

    # Output the db score
    print(f"  k = {k}: {dbScore:.6f}")
```

For a Silhouette score, the range is [-1,1], where a score of 1 is perfect clustering, 0 some data points overlap between clusters, and -1 being points are assigned to the wrong cluster. Therefore, a score closer to 1 is preferred. From the calculations, k = 2 produces the best clustering of points, with a silhouette score of 0.275. Yet again, this is not consistent with the ground truth for the number of clusters in the fashion_mnist dataset. (20). Usually, DBSCAN takes longer than K-means to run, and the time it takes to run is affected by the eps parameter. Explain why that is the case. [4 marks]

Answer to (20)

`eps` (epsilon) is the maximum radius around a given point and is used to determine which points are considered its neighbour. DBSCAN clusters points together that are within this distance threshold (bubble). Since eps is the threshold distance, generally the further you travel from a point, the more points you are going to interact with. This means that as the eps value gets larger, the DBSCAN algorithm will have more points to check against, increasing the time taken for the algorihtm to compute the clusters.

(34)	Discuss the advantages and disadvantages of using PCA, t-SNE, and UMAP for the visualization purpose, considering factors such as the preservation of local and global structures, computational efficiency, and the potential introduction of artifacts. [3 marks]

Answer to (34)

PCA (Principle Component Analysis) is very fast as it only involves linear algebra, it captures global structure, and is deterministic - same result every run. However, it can only capture linear relationships, so it often produces uninformative 2d graphs for non-linear data (introduces artifacts) - therefore poorly revealing clusters if the structure is non-linear.  
t-SNE (t-Distributed Stochastic Neighbor Embedding) is excellent at revealing local clusters, making clusters look extremely separated, and is great for visualising complex non-linear relationships/manifolds. However, it is much slower on larger datasets, it doesn't preserve gloabl structure (distances between clusters and cluster size are not meaningful), and can invent clusters.  
UMAP (Uniform Manifold Approximation & Projection) is much faster than t-SNE (often 5-20x faster) - so works well on large datasets, and captures both local and some global structures. However, it is still slower than PCA and global structures are still not perfectly reliable.

(35)	Explain how the choice between these dimensionality reduction techniques might change when, after visualising, you want to do the preprocessing for the machine learning task. [3 marks]

Answer to (35)

When choosing a dimensionality reduction algorithm, one that preserves the properties the clustering algorithm relies on is often preferred. PCA preserves global euclidean distances (linearly) and UMAP preserves local neighbourhoods reasonably well, these may improve clustering reliability and speed. Using K-Means together with PCA is common. t-SNE is great at creating visually beautiful, well-separated clusters, but these separations cannot be trusted quantitatively - as t-SNE forces data points into clusters, distorts global structure, may create clusters that don't even exist in the original data, and the size and distances between clusters are meaningless.

Answer to (36)

a) Dimensionality reduction alogorithms with low interpretability mean you are unable to explain/understand what the axes represent nor compare distances between clusters gloablly. Stability determines whether you can trust the patterns, clusters, or conclusions you draw from the embeddings. Therefore, you would prefer a dimensionality reduction algorithm with a high interpretability and high stability for input into a clustering algorithm (such as PCA) and an algorithm with low interpretability and stability for data visualisation, such as t-SNE.  
b) Both t-SNE and UMAP are stochastic algorithms, meaning they involve randomness during optimisation. Because of this, the exact shape, position, and separation of clusters can change every time you run them. Meaning one run could result in clearly separated clusters and the next run could display overlapping clusters. However, this can be mitigated, but not eradicated, using a random seed - 'random_state=42'.

(37).	In the context of classifying MNIST digits 7 and 8, explain why a classifier might struggle to distinguish between these digits even with high overall accuracy. Suggest two evaluation metrics that would better reveal this issue, and justify their use. [3 marks]

Answer to (37)

Sometimes poorly written 8s can look like a stylised 7s, and other times 7s with a crossbar can also look like 8s. Handwriting variability causes overlapping digits. This cause of the high overall accuracy is due to the majority of the digits in the MNIST dataset being easy to classify except for a few pairs, namely (4,9), (5,3), and (7,8). Overall there is a totla of 100 pairs of digits, so these tricky paris only account for 3%, meaning it is realy easy for a model to achieve ~99% accuracy but still struggle with these pairs. Your model could have 99% overall accuracy but only 60-70% efficiency on distinguishing 7 and 8s. A few metrics that try to reveal these few cases include per-class precision and recall scores, and a confusion matrix. By using per-class precision (e.g. of the predicted 7s, how many were correct?) and recall (e.g. of all true 7s, how many did the model find?), you will be immediately able to see if the digits 7 or 8 are disproportionately misclassified. A confusion matrix will directly expose class pairs that that model struggles with. 

(38). Given a classification task and a dataset, sometimes it’s impossible to make a classifier with 100% precision and 100% recall simultaneously. Explain why. [3 marks]

Answer to (38)

Precision = of the predictions for a particular class, how many were correct.  
Recall = of the true predictions for a particular class, how many did the model find?  
Consider the case were the model is able to correctly classify all digits of 5, from the MNIST dataset. This would mean that the model has a 100% recall score for the digit 5. However, this could mean that the model can also classify poorly written 3s as a 5, which would reduce the precision score for the digit 5. In this case, the variety of human handwriting means that some digits will almost certainly be misclassified by the model, meaning it could have 100% recall but 96% precision. 

(39). In a recommendation system for online shopping, a machine learning model achieves high precision but struggles with low recall when recommending products to users. Explain the implications of this trade-off for the user experience, and suggest a way to improve recall without significantly sacrificing precision. [4 marks]

Answer to (39)

Having a high precision and a low recall in a shopping recommendation algorithm would not be good for user experience. This low recall would mean that out of the hundreds of possible recommendations the user could receive, they only receive a few. But, this high precision means that out of these few correct recommendations, most are actually what the user would buy. Using this information, it suggests that the majority of recommendations the user would get will be something they are not interested in and not going to buy. This would not only be bad for user experience, but also for the reputation of the algorithm provider and profits of the retailers. A few methods to improve recall include; better feature representation, data augmentation, and class weighting. Using dimensionality reduction algorithms such as PCA can remove noise in the data and separate the clusters better, increasing recall. Data augmentation can increase recall by altering the data slightly but keeping general structure, increasing the amount of data points the model is trained on. Class weighting can increase recall by making the model treat mistakes in certain classes more important, forcusing its efforts to improve the recall of that class.

(40). In the context of classifying MNIST digits 9 and 0 for a postal code recognition system, discuss how overfitting might occur if the model is trained on a small, clean dataset. Propose a method to detect overfitting, and suggest a way to mitigate it. [4 marks]

Answer to (40)

Over fitting might occur when classifying MNIST digits 9 and 0, with a small, clean dataset due to the lack of variety in the data. Having a small, clean dataset means that the majority of 9s and 0s will be written a certain way meaning the model cannot classify 9s and 0s written in slightly different ways. The training data could only include 9s written with a curve at the bottom and 0s with a slash through the middle. Then the testing set could include 9s with no curve at the bottom and 0s with no slash. In this case, the model would be overfitted to certain style digits, rendering it useless with generalisation. A key sign that this is the case is a very high accuracy, precision and recall in the training set, but a dramtic decrease in these metrics in the test set. To mitigate this issue, the training dataset needs to include variety inorder to expose the model to more possibilities of digit style.
