```python
import numpy as np
import sympy as sp

def vec(P, Q):
    P = np.array(P, dtype=float)
    Q = np.array(Q, dtype=float)
    return Q - P

def dist(P, Q):
    return np.linalg.norm(vec(P, Q))

def angle_deg(u, v):
    u = np.array(u, dtype=float)
    v = np.array(v, dtype=float)
    cos = (u @ v) / (np.linalg.norm(u) * np.linalg.norm(v))
    cos = np.clip(cos, -1.0, 1.0)
    return np.degrees(np.arccos(cos)), cos
```

3) Use `safe_add` to test these incompatible pairs:
- \(A\in R^{2\times 3}\), \(B\in R^{3\times 2}\)
- \(A\in R^{2\times 3}\), \(B\in R^{2\times 4}\)

**Your code:**

```python
# TODO: implement the tasks in B1
# Set up the matrices
A = np.array([[ 1, 2,-2],
              [ 3, 0, 2]])
B = np.array([[ 4, 0,-4],
              [-1,-3,-4]])

# Q1: A + B
# Check whether A and B are additively conformable
# I.e. have the same number of rows and columns
if len(A) == len(B) and len(A[0]) == len(B[0]):
    print(f"1) A + B = \n{A + B}\n")

# Q2: write safe_add(A,B) to ensure A and B are additively conformable
def safe_add(A,B):
    # Check if A and B have the same shape
    if len(A) == len(B) and len(A[0]) == len(B[0]):
        return A + B
    else:
        # Inform the user that they don't have the same shape
        print("3) A and B are not additively conformable.\n")
        return False;

# Q3: Test safe_add(A,B)
A = np.ones((2,3))
B = np.ones((3,2))
result = safe_add(A,B)
if type(result) != bool:
    print(f"3) a) A + B = {result}\n")

A = np.ones((2,3))
B = np.ones((2,4))
result = safe_add(A,B)
if type(result) != bool:
    print(f"3) b) A + B = {result}\n")
```

Using random integer matrices (choose any sizes that make sense):

2) Verify that (A^TA) and (AA^T) are symmetric.

**Your code:**

```python
# TODO: generate random matrices and verify the identities
# Define matrix A and B with shape (2,3) and random values
A = np.array([[35, 4,16],
              [65,89,12]])
B = np.array([[34, 6,12],
              [ 5,73,42]])
A_transpose = np.transpose(A)
B_transpose = np.transpose(B)

# Q1: (A + B)^T = A^T + B^T
print(f"A = \n{A}\n\nA^T = \n{A_transpose}\n\nB = \n{B}\n\nB^T = \n{B_transpose}\n")
print("\nQ1)")
print(f"A + B = \n{A + B}\n")
print(f"(A + B)^T = \n{np.transpose(A + B)}\n")
print(f"A^T + B^T = \n{np.transpose(A) + np.transpose(B)}\n")
print("Therefore, (A + B)^T = A^T + B^T, for any shaped matrix.")

# Q2: (A^T)A = A(A^T)
print("\n\nQ2)")
print(f"(A^T) @ A = \n{A_transpose @ A}\n")
print(f"((A^T) @ A)^T = \n{np.transpose(A_transpose @ A)}\n")
print(f"A @ (A^T) = \n{A @ A_transpose}\n")
print(f"(A @ (A^T))^T = \n{np.transpose(A @ A_transpose)}\n")
print("Since each calculated matrix is equal to its transpose, they are both symmetric.")
```
