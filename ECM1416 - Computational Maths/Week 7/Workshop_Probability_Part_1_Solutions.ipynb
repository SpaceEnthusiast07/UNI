```python
import numpy as np
import matplotlib.pyplot as plt
```

# Q1

A manufacturing company produces 1,000,000 items and there are three machine for producing these items - machine A, machine B and machine C

Machine A produces 30% and the defect rate is 2%.
Machine B produces 50% and the defect rate is 3%. 
Machine C produces 20% and the defect rate is 5%.

Tasks
1. Given the total number of items, simulate the number of items produced by each machine with the given probabilities
2. Estimate that an item randomly selected is defective, i.e. estimate p(defective)
3. Estimate the conditional probability: p (machine B|defective)

```python
# Solution

N = 1000000

# 1. Given total number of items, simulate the number of items produced by each machine with the given probabilities
machines = np.random.choice(['A','B','C'], size = N, p=[0.3,0.5, 0.2])
plt.hist(machines)

plt.ylabel('Number of items')
plt.xlabel('Machine')
plt.show()

# 2: 
defect_probs = np.zeros(N)
defect_probs[machines == 'A'] = 0.02
defect_probs[machines == 'B'] = 0.03
defect_probs[machines == 'C'] = 0.05

defective = np.random.rand(N) <= defect_probs
print(defect_probs)

P_defective = defective.mean()

# 3.
P_B_given_defective = (np.sum((machines=='B')& defective) /
                      np.sum(defective)
                      )

print("Estimated P(defective):", P_defective)
print("Estimated P(B | defective):", P_B_given_defective)
```

# Q2
A random variable probability density function is defined with an Exponential distribution: $\lambda \exp^{(-\lambda x)}$, when $\ x\geq 0$ and 0 when $x < 0$. 

Tasks
1. Generate 500,000 random samples from the distribution. Use $\lambda=2$. You can use numpy to generate random samples from the distribution. 
2. For $x$ in the interval $[0,1]$, plot the probability density function. You can use x = np.linspace(0, 1, 1000) to generate values for x.
3. Following part 2 and using the same interval, estimate the empirical cumulative distribution function CDF (do not need to write or use the formula for CDF). Plot the empirical CDF against x.
4. Compute the empirical mean and variance from the simulated samples.

5. Compute the theoretical mean and variance of the distribution. Theoretical mean is given by $1/\lambda$ and theoretical variance is given by $1/\lambda^2$

```python
# Solution
# 1.
N = 500000
lambda_ = 2
samples = np.random.exponential(scale=1/lambda_, size=N)
# 2.
x_vals = np.linspace(0, 1, 100)
pdf_vals = lambda_ * np.exp(-lambda_ * x_vals)
plt.figure()
plt.plot(x_vals, pdf_vals)
plt.title("Exponential PDF (λ=2)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid()
plt.show()

# 3.
empirical_cdf = [np.mean(samples <= xi) for xi in x_vals]

plt.figure()
plt.plot(x_vals, empirical_cdf)
plt.title("Empirical CDF (λ=2)")
plt.xlabel("x")
plt.ylabel("F(x)")
plt.grid()
plt.show()

# 4.
empirical_mean = np.mean(samples)
empirical_variance = np.var(samples)

print("Empirical Mean:", empirical_mean)
print("Empirical Variance:", empirical_variance)

# 5.
theoretical_mean = 1 / lambda_
theoretical_variance = 1 / (lambda_ ** 2)

print("Theoretical Mean:", theoretical_mean)
print("Theoretical Variance:", theoretical_variance)
```
