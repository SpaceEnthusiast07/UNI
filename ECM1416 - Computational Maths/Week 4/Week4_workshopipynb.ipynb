```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)

def gaussian_elimination(Ab, tol=1e-10):
    """Gauss–Jordan elimination."""
    R = Ab.astype(float).copy()
    m, n = R.shape
    r = 0
    for c in range(n-1):
        # choose pivot row
        pivot = None
        max_val = tol
        for rr in range(r, m):
            v = abs(R[rr, c])
            if v > max_val:
                max_val = v
                pivot = rr
        if pivot is None:
            continue
        if pivot != r:
            R[[r, pivot]] = R[[pivot, r]]
        # normalize pivot row
        pv = R[r, c]
        R[r, :] /= pv
        # eliminate other rows
        for rr in range(m):
            if rr == r:
                continue
            factor = R[rr, c]
            if abs(factor) > tol:
                R[rr, :] -= factor * R[r, :]
        r += 1
        if r == m:
            break
    R[np.abs(R) < tol] = 0.0
    return R

def ranks(A, b, tol=1e-10):
    """(rank(A), rank([A|b])) via SVD numerical rank."""
    sA = np.linalg.svd(A, compute_uv=False)
    rA = int(np.sum(sA > tol))
    Ab = np.column_stack([A, b])
    sAb = np.linalg.svd(Ab, compute_uv=False)
    rAb = int(np.sum(sAb > tol))
    return rA, rAb

def classify_system(A, b, tol=1e-10):
    """Classify Ax=b as unique / infinite / none."""
    m, n = A.shape
    rA, rAb = ranks(A, b, tol=tol)
    if rAb > rA:
        return "no solution"
    if rA == n:
        return "unique solution"
    return "infinitely many solutions"
```
