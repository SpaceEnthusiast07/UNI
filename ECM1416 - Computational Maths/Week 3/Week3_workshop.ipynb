## Determinants and Inverses (Python/Jupyter)


```python
import numpy as np
import sympy as sp

np.set_printoptions(suppress=True, precision=6)

def is_identity(M, tol=1e-9):
    M = np.array(M, dtype=float)
    n = M.shape[0]
    return np.linalg.norm(M - np.eye(n), ord=np.inf) <= tol
```

## Exercise 3 — Singular vs invertible

Consider
\[
C_1 = [ 2&1&3\\ 4&2&6\\ 1&0&1 ]
C_2 = [ 2&1&3\\ 4&2&7\\ 1&0&1 ].
\]
1. Compute the determinants and decide which matrix is singular.
2. For the invertible one, compute its inverse and verify it.


```python
C1 = sp.Matrix([[2,1,3],[4,2,6],[1,0,1]])
C2 = sp.Matrix([[2,1,3],[4,2,7],[1,0,1]])

# Compute det(C1) and det(C2)
det_C1 = C1.det()
det_C2 = C2.det()

# Ouput the determinants
print(f"1)\n  det(C1) = {det_C1}\n  det(C2) = {det_C2}")
print("\n  Since the determinant of C1 is 0, it is singular and doesn't have an inverse.")
```

```python
# Compute the inverse for C2
C2_inverse = np.linalg.inv(np.array(C2, dtype=float))
print(f"2)\n  C2^(-1) = \n{C2_inverse}")

# Verify the inverse
# If inverse, C2 * C2^-1 = I
C2_C2_inverse = np.array(C2 @ C2_inverse, dtype=int)
print(f"  C2 @ C2_inverse = \n{C2_C2_inverse}\n  I = \n{np.eye(3, dtype=int)}")
if np.eye(3, dtype=int).all() == C2_C2_inverse.all():
    print("C2_inverse is the correct inverse.")
else:
    print("C2_inverse is not the inverse.")
```

## Exercise 6 — Solving a linear system with inverse vs solver

Solve the system Ax=b where
\[
A=[ 2&1&-1\\ 1&3&2\\ 3&1&1 ]
b=[ 2\\ 9\\ 7 ].
\]
1. Solve using `A.inv()*b` in `sympy` (exact rational solution).
2. Solve using `numpy.linalg.solve` (floating solution).
3. Confirm both solutions agree (within tolerance).


```python
A = sp.Matrix([[2,1,-1],[1,3,2],[3,1,1]])
b = sp.Matrix([2,9,7])
x = np.array(A.inv() @ b, dtype=float)
print(f"1)\n  x = A^-1 * b = \n{x}")
```

```python
# Numpy solve (float)
A_np = np.array(A.tolist(), dtype=float)
b_np = np.array(b.tolist(), dtype=float).reshape(-1)
x_np = np.linalg.solve(A_np, b_np).reshape((3, 1))
print(f"2)\n  x_np = A_np^-1 * b_np = \n{x_np}")
```

```python
# Compare (tolerance)
c = x_np - x
print(f"3)  x_np - x = \n{c}")
print(f"  Are they within tolerance? {c.all() <= 1e-9}")
```
