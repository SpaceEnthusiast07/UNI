# Q1 Modelling a Cooling Process (Newton's Law of Cooling)

Newton's Law of Cooling states that the rate of change of temperature of an object is proportional to the
difference between its temperature and the surrounding environment.

Mathematically, this is expressed as:

    dT/dt + kT = k * T_env,   where k > 0

where:
- T is the temperature of the object,
- t is time,
- T_env is the ambient temperature,
- k is a positive constant.

## Task:
Solve this ordinary differential equation (ODE) using the integrating factor method.

```python
# My Answer
import sympy as sp

# Define the symbols
T, t, T_env, k, C = sp.symbols('T t T_env k C')

# Define T as a function of t
T = sp.Function('T')(t)

# Calculate mu
mu = sp.exp(k*t)

# Multiply the right hand side of the ODE by mu
rhs = mu * k * T_env
# Now integrate the rhs
rhs = sp.integrate(rhs, t)

# Manually construct the trick of converting the lhs into the product rule format
lhs = mu * T

# Add the constant c onto the rhs and divide both sides by mu
T_solution = sp.simplify((rhs + C) / mu)
print(T_solution)
```

## Solution Using the Integrating Factor Method

The given ODE:

    dT/dt + kT = k * T_env

is a first-order linear differential equation of the form:

    dT/dt + p(t)T = q(t)

where p(t) = k and q(t) = k * T_env.

The integrating factor μ(t) is given by:

    μ(t) = exp(∫ p(t) dt) = exp(∫ k dt) = exp(kt).

Multiplying both sides of the ODE by e^(kt):

    e^(kt) * dT/dt + k * e^(kt) * T = k * T_env * e^(kt).

Rewriting the left-hand side:

    d/dt (T * e^(kt)) = k * T_env * e^(kt).

Integrating both sides,

    ∫ d/dt (T * e^(kt)) dt = ∫ k * T_env * e^(kt) dt.
We have,

    T * e^(kt) = 1/k * k * T_env * e^(kt) + C
Rearrange, the Solution is:

    T = T_env + C e^(-kt).




## Solution Using the Integrating Factor Method (Python Code)


```python
#import sympy as sp
# import sympy: This imports the SymPy library, a Python library for symbolic mathematics,
# which is useful for solving algebraic equations, calculus problems, and symbolic differentiation/integration.

# Define symbols
t, k, T_env, C = sp.symbols('t k T_env C')
T = sp.Function('T')(t)

# Define the integrating factor
mu = sp.exp(k*t)

# Multiply the ODE by the integrating factor
lhs = sp.Derivative(T * mu, t)
rhs = k * T_env * mu

# Integrate both sides
lhs_integrated = sp.integrate(lhs, t)
rhs_integrated = sp.integrate(rhs, t)

# Solve for T
T_solution = sp.simplify((rhs_integrated + C) / mu)
print(T_solution)
```
