### Question 1
Given the function  
\[
f(x, y) = x^3y + \sin(xy)
\]

1. Compute the partial derivatives \(\partial f / \partial x\) and \(\partial f / \partial y\).
2. Evaluate both partial derivatives at the point \((1,0)\).

You can use the sympy library for differentiation. Check the answer below


```python
from sympy import *

# Set up the algebraic symbols x and y
x, y = symbols('x y')

# Set up the expression, i.e. the function
expr = (x**3)*y + sin(x*y)

# Output the expression to the console
print(f"Expression: {expr}")

# Use the derivative method with respect to x
expr_diff_x = Derivative(expr, x)

# Use the derivative method with respect to y
expr_diff_y = Derivative(expr, y)

print("\n1)")
# Actually calculate the dertivative
# This calculates the partial derivative with respect to x
partial_x = expr_diff_x.doit()
partial_y = expr_diff_y.doit()
print(f"Evaluated dertivative with respect to x: {partial_x}")
print(f"Evaluated dertivative with respect to y: {partial_y}")

print("\n2)")
# Evaluate both derivatives at the point (1,0)
print(f"Partial x value: {partial_x.evalf(subs = {x:1.0, y:0.0})}")
print(f"Partial y value: {partial_y.evalf(subs = {x:1.0, y:0.0})}")
```

```python

# Solution Q1
import sympy as sp

x, y = sp.symbols('x y')
f = x**3*y + sp.sin(x*y)

fx = sp.diff(f, x)
fy = sp.diff(f, y)

fx, fy, fx.subs({x:1, y:0}), fy.subs({x:1, y:0})
```

\[
f(x, y) = x^2 + y^2 + xy
\]

1. Compute the gradient vector.
2. Evaluate the gradient at \((1,-1)\).
3. Plot contour lines of the function and overlay the gradient vector. For the contour plot, you will need to use meshgrid function.


```python
import numpy as np

# Set up the symbols used in the function
x, y = symbols('x y')

# Construct the function
expr = (x**2) + (y**2) + (x*y)

# Q1
# Since we are working in a 2 dimensional space (x & y),
# our gradient vector will be 2 dimensional.
# Calculate the partial derivatives
partial_diff_x = Derivative(expr, x).doit()
partial_diff_y = Derivative(expr, y).doit()

# Construct the gradient vector from the two partial derivatives above
gradient_vector = np.array([partial_diff_x, partial_diff_y])

# Output the gradient vector
print(f"Gradient vector: \n{gradient_vector}")
```
