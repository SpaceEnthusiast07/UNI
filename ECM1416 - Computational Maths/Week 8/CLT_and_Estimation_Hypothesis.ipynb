# Exercise 1: Central Limit Theorem SimulationTask: 

Create a population of 10,000 random integers between 0 and 100 (a Uniform distribution). Take 500 random samples from this population, each of size $n=40$. Calculate the mean of each sample. Plot a histogram of these 500 sample means.

Question: Does the histogram look uniform or normal? Calculate the mean of the sample means - how does it compare to the population mean?

```python
# Used to generate arrays and random integers
import numpy as np
# Used to plot the histogram of means
import matplotlib.pyplot as plt

def mean(listOfNumbers: list[int]) -> int:
    """Calculates the mean of a given array of integers.
    
    #### Keyword arguments:

    listOfNumbers -- The array of integers to calculate the mean of.

    #### Return: 
    
    An integer representing the mean of the array.
    """

    # Initialise the sum of integers
    sum = 0

    # Calculate the sum of the integers
    for i in range(len(listOfNumbers)):
        sum += listOfNumbers[i]
    
    # Now, return the mean of this array
    return sum / len(listOfNumbers)

# Initialise an array to hold all 10,000 random integers
allRandomIntegers = np.zeros(10_000)

# Use a loop to calculate 10,000 random integers between 0 and 100
for i in range(10_000):
    # Use the numpy randint function
    allRandomIntegers[i] = np.random.randint(0,101)

# Initialise an array to hold the random sample of size 40
randomSample = np.zeros(40)

# Initialise an array to hold the 500 sample means
sampleMeans = np.zeros(500, dtype=float)

# Loop 500 times to take 500 random samples of size 40 from this population
for i in range(500):
    # Choose 40 random samples from this population
    for j in range(40):
        # Generate a random index
        randomIndex = np.random.randint(0,10_000)
        # Store this random integer
        randomSample[j] = allRandomIntegers[randomIndex]
    
    # Calculate the mean of this sample
    sampleMeans[i] = mean(randomSample)

# Now, plot a histogram of these 500 sample means
# Graph Structure:
#  - The x-axis represents the sample number, i.e. sample 1, sample 2, sample 3, etc
#  - The y-axis represents the mean for that particular sample

# Configure figure output format
plt.figure(figsize=(10,6))
plt.title("Distribution of Sample Means from a Population of 10,000 Integers")
plt.xlabel("Sample Mean")
plt.ylabel("Sample Frequency")

# Set the x-axis and y-axis bounds
plt.xlim(0,100)
plt.ylim(0,25)

# Actually plot the histogram
plt.hist(x=sampleMeans, bins=100, color='green', alpha=0.7)

# Display the figure
plt.show()

# Calculate the mean od the samples
meanOfSamples = mean(sampleMeans)

# Calculate the mean of the population
populationMean = mean(allRandomIntegers)

# Output the two means
print(f"Population mean: {populationMean:.5f}")
print(f"Sample mean: {meanOfSamples:.5f}")
```

### Answer to Exercise 1
From the above diagram, you can see that the histogram follows the general shape of a normal distribution, which is expected for a large number of samples.

The sample mean and population mean are almost identical. Executing the program many times yeilds the information that the two means always stay within 1 unit of each other.
