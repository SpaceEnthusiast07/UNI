````python
from time import time

# Answer 2: Solving Linear Systems 
A = np.array([[12, -5,  3, -2],
              [-4, 15, -6,  7],
              [ 5, -3, 18, -9],
              [-2,  4, -8, 20]], dtype=float)

b = np.array([3, 7, 5, 9], dtype=float)

# You must not use numpy.linalg.solve, numpy.linalg.inv, or any external library function that performs matrix inversion or Gaussian elimination.
# (a) Solve the system using Gaussian elimination, implemented entirely by your own code. (10%)

# Start the timer
ge_start_timer = time()

# Augment matrix A with vector b
Ab_aug = np.ndarray((4,5))
# Glue vector b onto the right hand side of A
for i in range(4):
    Ab_aug[i] = np.append(A[i], b[i])

# Element A_11 is first pivot value
# R2 = R2 + (1/3)R1
Ab_aug[1] = Ab_aug[1] + (Ab_aug[0] / 3)
# R3 = R3 - (5/12)R1
Ab_aug[2] = Ab_aug[2] - ((5 * Ab_aug[0]) / 12)
# R4 = R4 + (1/6)R1
Ab_aug[3] = Ab_aug[3] + (Ab_aug[0] / 6)

# Element A_22 is next pivot value
# R3 = R3 + (A_32/A_22)R2
Ab_aug[2] = Ab_aug[2] - ((Ab_aug[2][1] * Ab_aug[1]) / Ab_aug[1][1])
# R4 = R4 - (A_42/A_22)R2
Ab_aug[3] = Ab_aug[3] - ((Ab_aug[3][1] * Ab_aug[1]) / Ab_aug[1][1])

# Element A_33 is the third pivot value
# R4 = R4 - (A_43/A_33)R3
Ab_aug[3] = Ab_aug[3] - ((Ab_aug[3][2] * Ab_aug[2]) / Ab_aug[2][2])

# A is now in Row Echelon Form (Upper-Triangular)!!

# Start backward substitution
x_4 = Ab_aug[3][4] / Ab_aug[3][3]
x_3 = (Ab_aug[2][4] - (Ab_aug[2][3] * x_4)) / Ab_aug[2][2]
x_2 = (Ab_aug[1][4] - (Ab_aug[1][3] * x_4) - (Ab_aug[1][2] * x_3)) / Ab_aug[1][1]
x_1 = (Ab_aug[0][4] - (Ab_aug[0][3] * x_4) - (Ab_aug[0][2] * x_3) - (Ab_aug[0][1] * x_2)) / Ab_aug[0][0]

# End the timer
ge_end_timer = time()

x_ge = np.transpose(np.array([x_1, x_2, x_3, x_4]))
print("Solution x (Gaussian elimination):")
print(x_ge)



# (b) Solve the system by computing the matrix inverse manually using your own implementation, and then compute x=A^−1 b. (5%)

def calc_minor(matrix: np.array, row: int, column: int) -> np.array:
    """
    This function calculates the minor of a given element, at position matrix_rowcolumn.
    
    Parameters:
     - row is 1 based
     - column is 1 based
    """

    # Convert row and column to 0-based
    row -= 1
    column -= 1

    # If matrix is 2x2, return single element
    if (len(matrix) == 2):
        return matrix[1][column+1]

    # Initialise the numpy array to hold the new matrix with the required row removed
    new_matrix_1 = []

    # Loop through each row and remove the required row
    for i in range(len(matrix)):
        if i != row:
            new_matrix_1.append(matrix[i])

    # Convert back to a numpy array
    new_matrix_1 = np.array(new_matrix_1)

    # Initialise a list to hold all elements with the required column removed
    element_list = []

    # Loop through each row
    for i in range(len(new_matrix_1)):
        # Loop through each column
        for j in range(len(new_matrix_1[0])):
            if j != column:
                element_list.append(new_matrix_1[i][j])
    
    # Convert to a np array and change the shape
    new_matrix_2 = np.array(element_list).reshape((len(new_matrix_1), len(new_matrix_1)))

    return new_matrix_2

def calc_determinant(matrix: np.array) -> int:
    """This is a recursive function that calculates the determinant for any size square matrix."""

    # Ensure the matrix is square
    if (len(matrix) != len(matrix[0])): raise ValueError(f"Matrix is not square. {len(matrix)} is not {len(matrix[0])}.")

    # If the matrix is 2x2, return the determinant
    if (len(matrix) == 2):
        return (matrix[0][0] * matrix[1][1]) - (matrix[0][1] * matrix[1][0])

    # Recursive section
    # Initialise the determinant
    determinant = 0

    # Use Lapace's Formula to calculate the determinant
    for column in range(len(matrix[0])):
        determinant += ((-1) ** (1 + column + 1)) * (matrix[0][column]) * (calc_determinant(calc_minor(matrix, 1, column + 1)))
    
    return determinant

# Start the timer
inv_start_timer = time()

# Use the above functions to calculate the determinant of matrix A
det_A = calc_determinant(A)

# Copy matrix A and then calculate the matrix of minors
matrix_of_minors = np.copy(A)

# Iterate through each element in A and calculate the minor for each
# At the same time, apply the sign net to calculate the matrix of cofactors
for row in range(len(matrix_of_minors)):
    for column in range(len(matrix_of_minors)):
        matrix_of_minors[row][column] = ((-1) ** (row + column)) * calc_determinant(calc_minor(A, row+1, column+1))

# Transpose the matrix of minors
matrix_of_minors_t = np.copy(matrix_of_minors)
for row in range(len(matrix_of_minors)):
    for column in range(len(matrix_of_minors)):
        temp = matrix_of_minors[column][row]
        matrix_of_minors_t[column][row] = matrix_of_minors[row][column]
        matrix_of_minors_t[row][column] = temp

# Divide the transpose of the matrix of cofactors by the determinant
A_inv = matrix_of_minors_t / det_A

# Multiply b by A^-1
x_inv = A_inv @ b

# End the timer
inv_end_timer = time()

print("\nSolution x (via A^{-1} b):")
print(x_inv)



# (c) Measure and compare the average running time and judge which method is more efficient. (5%)
t_ge = ge_end_timer - ge_start_timer
t_inv = inv_end_timer - inv_start_timer

print()
print(f"Average time (Gaussian elimination): {t_ge:.6e} seconds")
print(f"Average time (inverse):  {t_inv:.6e} seconds\n")

#Input (Gaussian elimination) or (Inverse method)
method = 'Gaussian elimination' if t_ge < t_inv else 'Inverse method'
print(method + " is more efficient.")
````
